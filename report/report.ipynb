We will test multiple probability distributions: Uniform distribution; Normal distribution; Gamma distribution and Log-normal distribution, and compare them with the Double Expectation Distribution.

For each dataset, we will:
- Plot histograms to visualize raw data distribution
- Fit multiple probability distributions
- Calculate statistical metrics for each distribution
