## Abstract

WeChat Red Envelopes have become an integral part of daily social interactions in China, especially during festive occasions like New Year's Day. 

This report explores the distribution patterns of individual red envelope amounts through a comprehensive analysis process. Our team first collected empirical data, including the scenario of 5 RMB for 3 people, as well as New Year's Day data to examine holiday-specific variations. 

We then visualized and analyzed both collected data. 

Next, we compared traditional statistical models (uniform, normal distributions) with our proposed double expectation distribution, evaluating their goodness-of-fit through various statistical tests while discussing the advantages and limitations of each testing approach. 

To expand our dataset and validate our findings, we employed generative diffusion models to generate synthetic samples and verified their alignment with the expected double expectation distribution. 

Additionally, we investigated optimal red envelope snatching strategies and explored further extensions, including analyzing distribution differences across holidays and formulating targeted strategies for different user groups. This multi-faceted approach provides both theoretical insights into red envelope distributions and practical recommendations for users.
